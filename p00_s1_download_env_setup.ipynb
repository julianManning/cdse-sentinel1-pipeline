## 🛠️ Conda Environment Lifecycle Manager
> ### **Environment Synchronisation → Production‑Ready Reproducibility**

Creates the `Sentinel1` environment if it does not exist, otherwise performs a **non‑destructive delta install** of any packages missing relative to the YAML definition.

---

### 📋 Behaviour & Limitations
* **Selective Installation**: Installs only missing packages; existing environments are never removed.
* **Conda-Only Scope**: `pip:` sections within YAML files are ignored.
* **Note:** It is recommended to restart your machine after the first installation ⚠️ so the Windows environment variables for GDAL are properly registered.

---

### 📋 Supported Environment Definitions
* 🔓 **`env.base.yml`**: Loose dependency specifications. Allows Conda to resolve the latest compatible versions.
* 🔒 **`env.lock.yml`**: Create one after a successful build with `conda env export -n Sentinel1 > Tools/env.lock.yml` for 1:1 reproducibility across machines.

```python
ensure_conda_environment(env_name="Sentinel1", env_yml_path="Tools/env.base.yml")
```

## 🛠️ Jupyter Kernel Registration
### **Environment Integration → Notebook Execution Readiness**

Maps the verified Conda environment to a dedicated kernel specification via `ipykernel`, making **"Python 3 (Sentinel1)"** available in the kernel picker.

---

### 📋 Execution Readiness
* **Kernel Switching**: Once registered, select the kernel in the top-right corner of the Jupyter interface.
* **Environment Isolation**: All libraries imported in subsequent notebooks are sourced strictly from the dedicated environment.

```python
setup_jupyter_kernel(env_name="Sentinel1")
```

## 🔍 Configuration Integrity Check
> ### **Environment & Kernel Validation → Final Deployment Assurance**

A **multi‑point diagnostic audit** that inspects the Jupyter `KernelSpecManager`, parses `kernel.json` and cross-references it with the local Conda registry.

---

### 📋 Diagnostic Troubleshooting
* **Kernel Not Found**: Environment exists, but registration was skipped.
* **Env Not Found**: Kernel specification exists, but the Conda environment was removed or renamed.
* **JSON Error**: The kernel path is corrupted or permissions are restricted.

```python
run_integrity_check(env_name="sentinel1")
```
