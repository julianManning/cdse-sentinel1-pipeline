# 📦 Sentinel-1 Data Provisioning Notebook: Overview
## **Automated Batch Extraction → Analysis-Ready .SAFE Structures**

This notebook serves as the dedicated data unpacking utility for the Sentinel-1 workflow. It automates the batch extraction of compressed `.zip` archives into their native `.SAFE` directories, preparing the raw SAR products for downstream processing (e.g. SNAP / `esa_snappy`, `pyroSAR`, ISCE or GAMMA).

---

### 🚀 Core Functional Areas
* **Automated Batch Decompression**: Scans the source directory and extracts all valid `.zip` archives to the target destination in a single execution.
* **Format Validation & Integrity**: Targets valid archives only and skips corrupted ones, producing the `.SAFE` folder structure expected by SAR toolchains.
* **Storage Lifecycle Management**: Optional cleanup removes the original `.zip` files post-extraction.

---

### 📋 Execution Requirements & Troubleshooting
* **Pipeline Sequencing**: Run after `p01_s1_download_pipeline.ipynb` has retrieved the compressed dataset.
* **SNAP can read zips directly**: SNAP / `esa_snappy` opens Sentinel-1 `.zip` archives without extraction. Unzip only if your downstream tooling needs the `.SAFE` folder.
* **Irreversible Deletion Warning**: With `delete_after_extraction=True` the original archives are permanently removed.
* **Cell State Configuration**: If your data is already unzipped, leave the execution cell as **'Raw'**.
